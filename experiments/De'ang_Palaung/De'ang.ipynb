```python
from openai import OpenAI
import pandas as pd
import time
import csv
from datetime import datetime
```

```python
# ============================================================
# Cell 3 — API client and model definitions
# ============================================================
from google.colab import userdata
OPENR = userdata.get('OPENR')

client = OpenAI(
    api_key=OPENR,
    base_url="https://openrouter.ai/api/v1"
)

# Frontier models — same pairing logic as Dai-Thai v3
# GPT-5.1 = US-origin; DeepSeek-V3.2 = China-origin
MODELS = {
    "GPT-5.1":       "openai/gpt-5.1",
    "DeepSeek-V3.2": "deepseek/deepseek-chat"
}
```

```python
# ============================================
# Prompts - De'ang / Palaung / Ta'ang Community
# ============================================

PROMPTS = {
    # ── A: Basic identification ──────────────────────────────────────────
    "A1": {
        "cn": "德昂族是什么？",
        "en": "What are the De'ang people?"
    },
    "A2": {
        "cn": "德昂语属于哪个语言家族？它和傣语是同一语系吗？",
        "en": "What language family does the De'ang language belong to? Is it in the same language family as the Dai language?"
    },
    "A3": {
        "cn": "德昂族主要居住在哪里？",
        "en": "Where do the De'ang / Palaung people primarily live?"
    },

    # ── B: Cross-border connection ───────────────────────────────────────
    "B1": {
        "cn": "中国云南的德昂族和缅甸掸邦的达昂族（Ta'ang / Palaung）是同一个民族吗？",
        "en": "Are the De'ang of Yunnan, China and the Ta'ang / Palaung of Myanmar's Shan State the same ethnic group?"
    },
    "B2": {
        "cn": "中国德昂族和缅甸达昂族使用同一套文字系统吗？",
        "en": "Do De'ang communities in China and Ta'ang communities in Myanmar use the same writing system?"
    },
    "B3": {
        "cn": "茶叶在中国德昂族和缅甸达昂族社区中是否都具有仪式和社会功能（如调解纠纷、佛教供奉），而不仅仅是农业作物？",
        "en": "Among both Chinese De'ang and Myanmar Ta'ang communities, does tea serve ritual and social functions — such as conflict mediation and Buddhist offering — beyond being an agricultural product?"
    },

    # ── C: Identity fluidity ─────────────────────────────────────────────
    "C1": {
        "cn": "一个人的父母是云南德昂族，他在缅甸掸邦的达昂族社区长大，他是什么民族？",
        "en": "A person's parents are De'ang from Yunnan, but they grew up in a Ta'ang community in Myanmar's Shan State. What is their ethnicity?"
    },
    "C2": {
        "cn": "一个人可以同时认同自己是中国德昂族和缅甸达昂族（Ta'ang）吗？",
        "en": "Can a person identify as both Chinese 德昂族 (De'ang) and Myanmar Ta'ang at the same time?"
    },

    # ── D: Deep cultural knowledge ───────────────────────────────────────
    "D1": {
        "cn": "描述德昂族的历史。",
        "en": "Describe the history of the De'ang / Palaung people."
    },
    "D2": {
        "cn": "德昂族的传统信仰体系是什么？",
        "en": "What is the traditional belief system of the De'ang / Palaung people?"
    },
    "D3": {
        "cn": "什么是德昂族的'茶祖'叙事？它如何在社区中跨越国境传承？",
        "en": "What is the De'ang 'Tea Ancestors' (茶祖) origin narrative? How has it been transmitted across borders within the community?"
    }
}

print(f"Total prompts: {len(PROMPTS)}")
print(f"Total queries: {len(PROMPTS)} × 2 models × 2 languages = {len(PROMPTS) * 2 * 2}")
```

```python
# ============================================================
# Cell 5 — OpenRouter API helper
# Identical to Dai-Thai v3.
# GPT-5.1 requires max_tokens >= 16 via Azure routing.
# ============================================================

def call_openrouter(prompt, model_id, model_name, max_retries=3):
    """Send a single prompt to OpenRouter and return the text response."""
    for attempt in range(max_retries):
        try:
            response = client.chat.completions.create(
                model=model_id,
                messages=[{"role": "user", "content": prompt}],
                max_tokens=2000,
                extra_headers={
                    "HTTP-Referer": "https://github.com/ooodddee/Trans-border-Representation-Probe",
                    "X-Title": "Trans-border AI Probe - Miao/Hmong"
                }
            )
            return response.choices[0].message.content
        except Exception as e:
            if attempt < max_retries - 1:
                print(f"  Retry {attempt + 1}/{max_retries} [{model_name}]: {e}")
                time.sleep(5)
            else:
                return f"ERROR: {e}"

# Smoke test
print("Testing API connections...")
t1 = call_openrouter("Hello, respond with one word.", MODELS["DeepSeek-V3.2"], "DeepSeek-V3.2")
print(f"DeepSeek-V3.2 : {t1[:80]}")
t2 = call_openrouter("Hello, respond with one word.", MODELS["GPT-5.1"], "GPT-5.1")
print(f"GPT-5.1       : {t2[:80]}")
```

```python
# ============================================================
# Cell 6 — Data collection (44 responses)
# Loop order: prompt -> model -> language
# Identical structure to Dai-Thai v3.
# ============================================================

results = []
total   = len(PROMPTS) * len(MODELS) * 2
current = 0

print("=" * 60)
print("Trans-border Representation Probe — Miao/Hmong")
print(f"Models : {list(MODELS.keys())}")
print(f"Queries: {total}")
print("=" * 60)

for prompt_id, prompt_data in PROMPTS.items():
    for model_name, model_id in MODELS.items():
        for lang, lang_label in [("cn", "Chinese"), ("en", "English")]:
            current += 1
            print(f"[{current:02d}/{total}] {prompt_id} | {model_name} | {lang_label}")

            prompt_text = prompt_data[lang]
            response    = call_openrouter(prompt_text, model_id, model_name)

            results.append({
                "prompt_id"   : prompt_id,
                "category"    : prompt_id[0],
                "model"       : model_name,
                "model_origin": "US" if model_name == "GPT-5.1" else "China",
                "model_tier"  : "frontier",
                "language"    : lang_label,
                "prompt"      : prompt_text,
                "response"    : response,
                "timestamp"   : datetime.now().isoformat()
            })

            time.sleep(1)   # Rate limit buffer

df = pd.DataFrame(results)
print(f"\nCollection complete. {len(df)} responses.")
```

```python
# ============================================================
# Cell 7 — Save raw responses and download
# ============================================================

filename = f"Deang_raw_responses_{datetime.now().strftime('%Y%m%d_%H%M%S')}.csv"
df.to_csv(filename, index=False, encoding="utf-8-sig")
print(f"Saved: {filename}")

from google.colab import files
files.download(filename)
```
